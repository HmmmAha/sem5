```python
import librosa
import soundfile as sf
from pydub import AudioSegment
from IPython.display import display, Audio
import torch, torchaudio
import numpy as np
```

```python
AUDIO_FILE = './speech.wav'
```

```python
def volume():
    audio = AudioSegment.from_wav(AUDIO_FILE)

    db = 10

    audio_adjust = audio + db

    audio_adjust.export('volume.wav', format="wav")
```

```python
def pitch():
    audio, sr = librosa.load(AUDIO_FILE)

    pitch = 5

    audio_adjust = librosa.effects.pitch_shift(audio, sr=sr, n_steps=pitch)

    sf.write('pitch.wav', audio_adjust, sr)
```

```python
def fade_in():
    audio = AudioSegment.from_wav(AUDIO_FILE)

    secs = 5

    audio_fade_in = audio.fade_in(secs * 1000)

    audio_fade_in.export('fade_in2.wav', format="wav")
```

```python
def fade_out():
    audio = AudioSegment.from_wav(AUDIO_FILE)

    secs = 5

    audio_fade_out = audio.fade_out(secs * 1000)

    audio_fade_out.export('fade_out2.wav', format="wav")
```

```python
bundle = torchaudio.pipelines.WAV2VEC2_ASR_BASE_960H
model = bundle.get_model()
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
```

```python
def predict_text():
    audio, sr = torchaudio.load(AUDIO_FILE)

    if sr != bundle.sample_rate:
        torchaudio.functional.resample(audio, sr, bundle.sample_rate)

    with torch.inference_mode():
        emission, _ = model(audio)

    labels = bundle.get_labels()

    idx = torch.argmax(emission[0], dim=-1)
    idx = torch.unique_consecutive(idx, dim=-1)
    idx = [i for i in idx if i != 0]

    pred = "".join(labels[i] for i in idx)

    print(pred.replace("|", " "))
```

```python
def repeat():
    audio = AudioSegment.from_wav(AUDIO_FILE)

    repeat = 5

    repeat_audio = audio * repeat

    repeat_audio.export('hhe.wav', format='wav')
```

```python
def add_white_noise():
    audio, sr = librosa.load(AUDIO_FILE, sr=None)
    white_noise = np.random.normal(0, audio.std(), len(audio))

    white_noise_scale = 0.5

    white_noise_audio = audio + white_noise * white_noise_scale

    sf.write('whte.wav', white_noise_audio, sr)
```
