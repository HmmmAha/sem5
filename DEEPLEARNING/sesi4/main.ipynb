```python
# preprocessing
def preprocessing(word: str):
    word = word.lower()
    word = word.r.sub(r"([?.!,])", r"\1", word)
    word = word.r.sub(r'[" "]+', " ", word)
    word = word.r.sub(r'[^a-zA-Z?!.,]+', " ", word)
    word = word.strip()
    word = '<start>' + word + '<end>'
    
    return word

# print(preprocessing("Hi.	Hai.	CC-BY 2.0 (France) Attribution: tatoeba.org #538123 (CM) & #11252770 (Pacific)"))



def load_dataset(path, num_examples=None):
    with open(path, "r", encoding='utf-8') as file:
        lines = file.read().strip().split("\n")
        
    parts = [line.split('\t') for line in lines[:num_examples]]
    
    input_data = [preprocessing(part[0]) for part in parts]
    target_data = [preprocessing(part[1]) for part in parts]
    
    return input_data, target_data


file_path = './ind.txt'
input_data, target_data = load_dataset(file_path, num_examples=100000)

print(input_data)
print(target_data)

```
